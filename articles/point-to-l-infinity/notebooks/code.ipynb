```python
import matplotlib.pyplot as plt





# Points A and Bee
A = (1, 1)
B = (4, 5)

# Create the plot
plt.figure(figsize=(6, 6))
plt.plot([A[0], B[0]], [A[1], B[1]], 'ko')  # plot points
plt.text(A[0], A[1], 'A', fontsize=12, ha='right')
plt.text(B[0], B[1], 'B', fontsize=12, ha='right')

# Draw Manhattan path (L-shape)
plt.plot([A[0], B[0]], [A[1], A[1]], 'r--')  # horizontal leg
plt.plot([B[0], B[0]], [A[1], B[1]], 'r--')  # vertical leg

# Optional: add arrows
plt.annotate('', xy=(B[0], A[1]), xytext=A,
             arrowprops=dict(arrowstyle='->', color='red'))
plt.annotate('', xy=B, xytext=(B[0], A[1]),
             arrowprops=dict(arrowstyle='->', color='red'))

# Show grid
plt.grid(True)
plt.xlim(0, 6)
plt.ylim(0, 6)
plt.gca().set_aspect('equal')
plt.title('Manhattan Distance from A to B')
plt.show()
```

```python
import matplotlib.pyplot as plt

# List of points you want to show Manhattan distance for
points = [(1, 4), (3, 1), (5, 2), (-2, 5), (4, -3)]
origin = (0, 0)

fig, ax = plt.subplots(figsize=(6, 6))

# Plot origin and each point
ax.plot(*origin, 'ko', label='Origin')
for i, p in enumerate(points):
    ax.plot(p[0], p[1], 'bo')
    ax.text(p[0], p[1], f"P{i+1}{p}", fontsize=9, va='bottom', ha='right')

# Draw Manhattan (L-shaped) connectors
for p in points:
    # horizontal leg from (0,0) to (p.x, 0)
    ax.plot([origin[0], p[0]], [origin[1], origin[1]], 'r--', linewidth=1)
    # vertical leg from (p.x, 0) to (p.x, p.y)
    ax.plot([p[0], p[0]], [origin[1], p[1]], 'r--', linewidth=1)
    # annotate distance components
    dx = abs(p[0] - origin[0])
    dy = abs(p[1] - origin[1])
    ax.text(p[0]/2, origin[1] + 0.1, f"{dx}", color='r', fontsize=8, ha='center')
    ax.text(p[0] + 0.1, p[1]/2, f"{dy}", color='r', fontsize=8, va='center')

# Styling
ax.set_title("Manhattan Distance from Origin to Multiple Points")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.grid(True)
ax.set_aspect('equal', 'box')
ax.legend()

plt.show()
```

```python
import numpy as np

vectors = np.array([
    [3, 1, 1],
    [-1,3, 1],
    [-3, -4, 1],
    [-1.5, -3, -1],
    [1, -1, -2]
])

import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401
import seaborn as sns

# --- Global Style & Palette ---
sns.set_style("white")
sns.set_palette("bright")
plt.rcParams.update({
    "grid.color":    "lightgray",
    "grid.linestyle":"--",
    "grid.linewidth":0.7,
})




# --- Prepare colors for Manhattan paths ---
manhattan_colors = sns.color_palette("bright", len(vectors))

# --- Create figure and 3D axes ---
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
ax.set_facecolor('whitesmoke')

# --- Plot points without labels ---
ax.scatter(vectors[:, 0], vectors[:, 1], vectors[:, 2],
           s=100, edgecolor='black', c=manhattan_colors)
ax.scatter(0, 0, 0, color='black', marker='X', s=100)

# --- Draw Manhattan paths (solid colored) ---
for i, (v, color) in enumerate(zip(vectors, manhattan_colors)):
    x, y, z = v
    path = np.array([[0, 0, 0],
                     [x, 0, 0],
                     [x, y, 0],
                     [x, y, z]])
    ax.plot(path[:, 0], path[:, 1], path[:, 2],
            color=color, linewidth=2,
            label='Manhattan path' if i == 0 else None)

# --- Draw Euclidean paths (dashed black) ---
for i, v in enumerate(vectors):
    x, y, z = v
    ax.plot([0, x], [0, y], [0, z],
            color='black', linestyle='--', linewidth=2,
            label='Euclidean path' if i == 0 else None)

# --- Enable and style the grid on all three planes ---
ax.grid(True)
for axis in (ax.xaxis, ax.yaxis, ax.zaxis):
    axis._axinfo['grid']['color']     = plt.rcParams["grid.color"]
    axis._axinfo['grid']['linestyle'] = plt.rcParams["grid.linestyle"]
    axis._axinfo['grid']['linewidth'] = plt.rcParams["grid.linewidth"]

# --- Labels, Title, Legend ---
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Manhattan and Euclidean Distances in 3-Dimensional Space')
ax.legend(frameon=False)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Define points A and B
A = (0, 0)
B = (3, 2)

# Coordinates for the Manhattan path
x_manhattan = [A[0], B[0], B[0]]
y_manhattan = [A[1], A[1], B[1]]

# Create plot
fig, ax = plt.subplots(figsize=(6, 6))

# Plot Euclidean distance
ax.plot([A[0], B[0]], [A[1], B[1]], label='Euclidean distance')

# Plot Manhattan distance
ax.plot(x_manhattan, y_manhattan, label='Manhattan distance')

# Scatter points A and B
ax.scatter([A[0], B[0]], [A[1], B[1]])

# Annotate points
ax.text(A[0], A[1]+0.05, ' A', verticalalignment='bottom', horizontalalignment='right')
ax.text(B[0]-0.15, B[1]-.02, ' B', verticalalignment='bottom', horizontalalignment='left')

# Annotate v1 and v2 with v1 moved even closer to the axis
ax.text((A[0] + B[0]) / 2, A[1] + 0.2, '$v_1$', verticalalignment='top', horizontalalignment='center')
ax.text(B[0] - 0.2, (A[1] + B[1]) / 2, '$v_2$', verticalalignment='center', horizontalalignment='left')

# Formatting
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_aspect('equal', 'box')
ax.grid(True)
ax.legend()

plt.show()
```

```python
import random

# 1) Make 10 vectors of lengths 1 to 10
vectors = [
    [random.randint(0, 100) for _ in range(length)]
    for length in range(1, 11)
]

# 2) Store each vector’s largest entry
largest_entries = [max(vec) for vec in vectors]

# Show results
for i, (vec, mx) in enumerate(zip(vectors, largest_entries), start=1):
    print(f"Vector #{i} (len={len(vec)}): {vec}")
    print(f"  → Largest entry: {mx}\n")
```

```python
import matplotlib.pyplot as plt
import random

# Generate 10 vectors of lengths 1 to 10, random integers between -50 and 50
vectors = [[random.randint(-50, 50) for _ in range(n)] for n in range(1, 11)]
vectors9 = vectors[:9]

# p values to evaluate
p_values = [1, 2, 3, 5, 10, 20, 50, 100, 1000]

# Compute largest absolute entry for each vector
largest_abs = [max(abs(x) for x in vec) for vec in vectors9]

# Create a 3x3 grid of subplots
fig, axes = plt.subplots(3, 3, figsize=(12, 12))

for ax, vec, lb in zip(axes.flat, vectors9, largest_abs):
    # Scale to avoid overflow for large p
    if lb == 0:
        norms = [0] * len(p_values)
    else:
        ratios = [abs(x) / lb for x in vec]
        norms = [lb * sum(r**p for r in ratios)**(1/p) for p in p_values]

    # Plot bar chart with orange bars
    ax.bar([str(p) for p in p_values], norms, color='orange', zorder=2)

    # Add horizontal line for the largest absolute entry, plotted on top
    ax.axhline(lb, linestyle='--', color='black', linewidth=2, zorder=3)

    # Add a thin grid behind everything
    ax.grid(True, linestyle='--', linewidth=0.5, color='gray', zorder=1)

    # Annotate subplot
    ax.set_title(f'Vector length = {len(vec)}')
    ax.set_xlabel('p')
    ax.set_ylabel('Lᵖ norm')
    ax.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```python
from sklearn.datasets import make_regression
from sklearn.linear_model import Lasso, Ridge

X, y = make_regression(n_samples=100, n_features=30, n_informative=5, noise=10)

model = Lasso(alpha=0.1).fit(X, y)
print("Lasso nonzero coeffs:", (model.coef_ != 0).sum())

model = Ridge(alpha=0.1).fit(X, y)
print("Ridge nonzero coeffs:", (model.coef_ != 0).sum())
```
