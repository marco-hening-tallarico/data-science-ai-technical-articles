# Hot Rod Simulation (Runnable Environment Version)

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 1, 101)
temperature = np.exp(-((x-0.5)**2)/(2*0.08**2))
temperature += 0.15*np.sin(2*np.pi*x)
fig, ax = plt.subplots(figsize=(8,4))
ax.plot(x, temperature, lw=2, color='tab:red')
ax.set_title('Initial Temperature Profile (Hot Rod)')
ax.set_xlabel('Position x')
ax.set_ylabel('Temperature')
ax.grid(alpha=0.3)
fig.tight_layout()
out_dir = Path('../figures')
out_dir.mkdir(parents=True, exist_ok=True)
fig_path = out_dir / 'hot_rod_temperature_profile.png'
fig.savefig(fig_path, dpi=200)
print('Saved:', fig_path)
plt.show()
```
